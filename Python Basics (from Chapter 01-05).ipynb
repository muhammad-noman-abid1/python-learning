```python
# Take the first number from the user
num1 = float(input("Enter first number: "))

# Ask the user which mathematical operation they want
operator = input("Enter operator (+, -, *, /): ")

# Take the second number from the user
num2 = float(input("Enter second number: "))


# Check if the operator is addition
if operator == "+":
    result = num1 + num2

# Check if the operator is subtraction
elif operator == "-":
    result = num1 - num2

# Check if the operator is multiplication
elif operator == "*":
    result = num1 * num2

# Check if the operator is division
elif operator == "/":

    # Prevent division by zero
    if num2 != 0:
        result = num1 / num2
    else:
        result = "Error: Cannot divide by zero"

# If the user enters an invalid operator
else:
    result = "Error: Invalid operator"


# Display the final result
print("Result:", result)
```

```python
# sets
a={1,2,3,4,4,4,5,"nomi","rohan","ali"}
b={4,5,6,7,8,"patlu","rohan","motu",}
print(a.union(b)) # get one time from sets
print(a.intersection(b)) # find common from both
print(a.difference(b)) # which elements are diff in A from B
print(b.difference(a)) # which elements are diff in B from A
a.add("ahmed") # for adding any element
print(a)
pa= (a.pop()) # for Removes and returns an arbitrary element , Don't assume pop() removes the first or last element. Sets are unordered.
print(pa)
print(a.issubset(b)) # Checks foe all elements of one set exist in another.
print(a.issuperset(b)) # check for  all elements of one set exist in another.
print(a.isdisjoint(b)) # Checks whether two sets have a null intersection.
```
