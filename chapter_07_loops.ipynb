06_ for loop with else

07_ break and continue .py

```python
# continue for skip the iteration and break for skipping loop
for i in range(40):
  if i == 15:
    break # break means exit the loop right now
  print(i)
for i in range(40):
  if i == 15:
    continue # used inside a loop to skip the current iteration and immediately move to the next iteration.
    # ee in put put 15 is not printed
  print(i)
```

08_ pass.py

```python
# pass statement means  do nothing here, but don't give an error
# use when you want to create a block of code now but implement it later
#  Without pass, an empty function would cause an error
for i in range (645):
  pass

i = 0
while(i<20):
  print(i)
  i +=1
```

Q4- find whether a given number is prime or not

```python
n=int(input("Enter a number: "))
for i in range(2,n):
  if(n%i)==0:
    print ("number is not prime")
    break
else :
      print("number is prime")
```

Q5 -using while loop , find sum of all the numbers before that given number

```python
n = int(input(" Enter a number: "))
i = 1
sum = 0
while(i<=n):
  sum += i
  i += 1
print(f"sum of first {n} numbers is {sum}")
```

Q6 - find factorial of given number using for loop

```python
#  5! = 1x2x3x4x5
n=int(input("Enter a number: "))
product=1
for i in range(1,n+1):
  product= product*i
print(f"factorial of {n} is {product}")
```

Q7-  print the given star pattren
"  *  
 ***
*****  "

```python
''' this star pattern
   *
  ***
 *****
*******
'''
n= int(input("Enter the number: "))
for i in range (1,n+1):
  print (" "*(n-i),end="") # here it will print the spaces of sequence, print always give new line by default but using "end" it not move to new line
  print("*"* (2*i-1),end="") # here it will print the stars of sequence
  print("") # here it will print one space in front of star and move to new line, means it will move to new line
```
