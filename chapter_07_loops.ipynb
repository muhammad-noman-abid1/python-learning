04_for_loops.py

```python
for i in range (4): # range funtion is used to generate the sequence of number , also specify start or stop at , or step size
  print(i)

for i in range(4,40,4):  # a for loop generate a sequence of numbers with a specified step size.
  print(i)
```

05_for_loop_iterate.py

```python
l=[1,3,6,534,6] # for loopin lists

for i in l:
  print(i)


t = (1,3,543,3,) # for loop can be used in tuples

for i in t:
  print(i)

s=("nomi") # can be used to iterate string/"str"
for i in s:
  print(i)
```

06_ for loop with else

```python
l=[1,2,5]
for item in l:
  print(item)

else:
  print("done") # this will print when the loop gets exhaust means when it wil complete the for loop first
```

07_ break and continue .py

```python
# continue for skip the iteration and break for skipping loop
for i in range(40):
  if i == 15:
    break # break means exit the loop right now
  print(i)
for i in range(40):
  if i == 15:
    continue # used inside a loop to skip the current iteration and immediately move to the next iteration.
    # ee in put put 15 is not printed
  print(i)
```

08_ pass.py

```python
# pass statement means  do nothing here, but don't give an error
# use when you want to create a block of code now but implement it later
#  Without pass, an empty function would cause an error
for i in range (645):
  pass

i = 0
while(i<20):
  print(i)
  i +=1
```
