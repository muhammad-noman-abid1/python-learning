```python
i = 1

while (i < 6):
    print(i)
    i += 1
# this loop is run again and again until condition is satisfief like ,it will print 1,2,3,4,5 because contion is satisfied till 5 and then it will stops because 6 is not grater than 6 so it will exit the loop at 6'''
```

04_for_loops.py

```python
for i in range (4): # range funtion is used to generate the sequence of number , also specify start or stop at , or step size
  print(i)

for i in range(4,40,4):  # a for loop generate a sequence of numbers with a specified step size.
  print(i)
```
