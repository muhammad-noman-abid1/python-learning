05- Recursions.py

```python
'''

factorial(1) = 1
factorial(2) = 2 X 1
factorial(3) = 3 X 2 X 1
factorial(4) = 4 X 3 X 2 X 1
factorial(5) = 5 X 4 X 3 X 2 X 1
factorial(n) = n X (n-1) X (n-2) X (n-3) X (n-4) ... X 1

factorial(n) = n * factorial(n-1) , the condition

'''
def factorial(n): # first i define a function to calculate the factorial of n
  if (n == 1 or n == 0):
    return 1    # if condition of if is true then stop the recursion and return 1 , otherwise go ahead
  return n * factorial(n-1)   # multiply n by the factorial of the previous number

n= int(input("Enter a number "))
print(f"Factorial of given number is '{factorial(n)}'") # here call the function and print the calculated value
```

Q1- using function to find the gratest of 3 numbers

```python
a= int(input("enter first number: "))
b= int(input("enter second number: "))
c= int(input("enter third number: "))
def great(a,b,c):
  if (a>b and a>c):
    print(f"{a} is the gratest number")
  elif (b>a and b>c):
    print(f"{b} is the gratest number")
  elif(c>a and c>b):
    print(f"{c} is the gratest number")

great(a,b,c)
```

Q2- convert Q2- convert fahrenheit to celcius

```python
def F_to_C(f):
  return 5*(f-32)/9
f=float(input("Enter the temperature in fahrenheit: "))
c=F_to_C(f)  # for making the code easy and readable
print(f"{round(c, 2)}°C")   # rounded the result to 2 decimal places and print it
```

Q4- use recurcive funtion to calculate the sum of first n natural number

```python
'''
logic

sum(1) = 1
sum(2) = 1 + 2
sum(3) = 1 + 2 + 3
sum(4) = 1 + 2 + 3 + 4
sum(5) = 1 + 2 + 3 + 4 + 5

sum(n) = 1 + 2 + 3 + 4 + 5 ........+(n-1) + n
sum(n) = sum(n-1) + n
'''
def sum(n):
  if(n==1):  # this base condition is used to prevent the code running to infinite loop even in backwards , and to prevent the code from crashing
    return 1
  return sum(n-1)+n
n=int(input("Enter a number: "))
print(f"The sum of all numbers till {n} is '{sum(n)}'")
```

Q5- to print first n lines of following pattern

```python
def pattern(n):
  if (n==0):
    return
  print("*" * n)
  pattern(n-1)
n=int(input("Enter how many times you want the pattern: "))

pattern(n)
```
