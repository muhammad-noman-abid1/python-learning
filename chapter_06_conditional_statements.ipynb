01_ conditional.py

```python
# @title
a=int(input("Enter your age: "))
if(a>=18): #  if the condition of IF is true then else will not run
  print("you are eligible for voting") # the space in the starting means that we are in the if , called intendation
  print("Thank you")
else:
  print("you are not eligible for voting") # if the condition of IF is false then else will run

```

02_ if , elif , else , ladder .py

```python
# @title
a=int(input("Enter your age: "))
# this process is called : if , elif , else , ladder
if(a>=18): #  if the condition of IF is true then elif and else will be ignore
  print("you are eligible for voting") # the space in the starting means that we are in the if , called intendation
  print("Thank you")

elif(a<0): # use for including multiple conditions
  print("you are entering wrong age ")
  print("please enter age again")
elif(a==0):
  print("why you are entering zero")
  print("please enter age again")
else:
  print("you are not eligible for voting") # if the condition of IF is false then else will run

print("ended!")
```

03_ multiple if statements

```python
a=int(input("Enter your age: "))
if(a%2 == 0): # if statement 1 , and this if statement is independent
  print("the number is even")
# is statement 1 ends here
if(a>=18): # if statement 2
  print("you are eligible for voting")
  print("Thank you")

elif(a==0):
  print("why you are entering zero")
  print("please enter age again")
else:
  print("you are not eligible for voting") # else never run alone without if statement
# end of if statement 2
print("ended!")
```

Q6- calculate the grade of student from his marks

```python
marks=int(input("Enter your marks: "))
if(marks<=100 and marks>=90):
  grade="Excellent"
elif(marks>=80 and marks<=89):
  grade="A"
elif(marks>=70 and marks<=79):
  grade="B"
elif(marks>=60 and marks<=69):
  grade="C"
elif(marks>=50 and marks<=59):
  grade="D"
elif(marks<50):
  grade="F"
  print("Fail")
print("Your grade is: ",grade)
```
