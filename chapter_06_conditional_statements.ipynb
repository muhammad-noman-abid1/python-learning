01_ conditional.py

```python
a=int(input("Enter your age: "))
if(a>=18): #  if the condition of IF is true then else will not run
  print("you are eligible for voting") # the space in the starting means that we are in the if , called intendation
  print("Thank you")
else:
  print("you are not eligible for voting") # if the condition of IF is false then else will run

```

02_ if , elif , else , ladder .py

```python
a=int(input("Enter your age: "))
# this process is called : if , elif , else , ladder
if(a>=18): #  if the condition of IF is true then elif and else will be ignore
  print("you are eligible for voting") # the space in the starting means that we are in the if , called intendation
  print("Thank you")

elif(a<0): # use for including multiple conditions
  print("you are entering wrong age ")
  print("please enter age again")
elif(a==0):
  print("why you are entering zero")
  print("please enter age again")
elif(a>100):
  print("you are dead")
else:
  print("you are not eligible for voting") # if the condition of IF is false then else will run

print("ended!")
```
